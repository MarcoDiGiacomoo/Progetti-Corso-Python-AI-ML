```python
#- Creare una lista con almeno 5 titoli di libri.
#- Creare un dizionario che mappi il titolo del libro al numero di copie disponibili.
#- Creare un insieme (set) che contenga tutti gli utenti registrati alla biblioteca.

libri = ["1Q84", "Norwegian Wood", "Una Vita Come Tante", "22.11.63", "Atti Umani", "La Vegetariana", "It"]

diz = {libri[0]: 12, libri[1]: 9, libri[2]: 4, libri[3]: 8, libri[4]: 14, libri[5]: 11, libri[6]: 17}
utenti = ("Marco", "Laura", "Luca", "Sara", "Silvia")

print(diz, type(diz))
print(utenti, type(utenti))
```

```python
#*Parte 3 – Classi e OOP*
#- Creare una classe Libro con attributi: titolo, autore, anno, copie_disponibili
#- Aggiungere un metodo info() che restituisca una stringa descrittiva del libro.
#- Creare una classe Utente con attributi:nome, eta, id_utente
#- Aggiungere un metodo scheda() che stampi i dati dell’utente.
#- Creare una classe Prestito che colleghi un Utente a un Libro e contenga: utente (oggetto Utente), libro (oggetto Libro), giorni (numero di giorni del prestito)
#- Aggiungere un metodo dettagli() che stampi tutte le informazioni sul prestito.

class libro:
    def __init__(self, titolo, autore, anno, copie_disponibili):
        self.titolo = titolo
        self.autore = autore
        self.anno = anno
        self.copie_disponibili = copie_disponibili

    def info(self):
        print(f"Titolo: {self.titolo}, Autore: {self.autore}, Anno: {self.anno}, Copie: {self.copie_disponibili}")

class utente:
    def __init__(self, nome, eta, id_utente):
        self.nome = nome
        self.eta = eta
        self.id_utente = id_utente

    def scheda(self):
        print(f"Nome: {self.nome}, Età: {self.eta}, ID Utente: {self.id_utente}")

class prestito:
    def __init__(self, libro, utente, giorni):
        self.libro = libro
        self.utente = utente
        self.giorni = giorni

    def dettagli(self):
        print(f"Titolo: {self.libro.titolo}, Utente: {self.utente.nome}, Giorni di Prestito: {self.giorni}")


lbr = libro("Kafka Sulla Spiaggia", "Murakami Haruki", 2002, 14)
utn = utente("Marco", 26, 125)
prs = prestito(lbr, utn, 30)

lbr.info()
utn.scheda()
prs.dettagli()
```

```python
#Parte 4 – Funzionalità*
#- Creare una funzione presta_libro(utente, libro, giorni) che:
#- Verifichi se il libro ha almeno 1 copia disponibile 
#     Se sì → riduca il numero di copie e crei un nuovo oggetto Prestito
#     Se no → stampi un messaggio di errore
#- Simulare almeno 3 prestiti con utenti e libri diversi.
#- Stampare a video:
#    1. L’elenco aggiornato delle copie disponibili per ciascun libro
#    2. I dettagli di ogni prestito effettuato

class libro:
    def __init__(self, titolo, autore, anno, copie_disponibili):
        self.titolo = titolo
        self.autore = autore
        self.anno = anno
        self.copie_disponibili = copie_disponibili

    def info(self):
        print(f"Titolo: {self.titolo}, Autore: {self.autore}, Anno: {self.anno}, Copie Disponibili: {self.copie_disponibili}")

class utente:
    def __init__(self, nome, eta, id_utente):
        self.nome = nome
        self.eta = eta
        self.id_utente = id_utente

    def scheda(self):
        print(f"Nome: {self.nome}, Età: {self.eta}, ID Utente: {self.id_utente}")

class prestito:
    def __init__(self, utente, libro, giorni):
        self.utente = utente
        self.libro = libro
        self.giorni = giorni

    def dettagli(self):
        print(f"Nome: {self.utente.nome}, Libro: {self.libro.titolo}, Giorni: {self.giorni}")

def presta_libro(utente, libro, giorni):
    if libro.copie_disponibili == 0:
        raise ValueError(f"Non ci sono copie disponibili di: {libro.titolo}")
    
    else:
        libro.copie_disponibili -= 1
        nuovo_prestito = prestito(utente, libro, giorni)
        return nuovo_prestito


utn1 = utente("Marco", 26, "125")
utn2 = utente("Silvia", 23, "099")
utn3 = utente("Lucia", 28, "050")

lbr1 = libro("Kafka Sulla Spiaggia", "Murakami Haruki", 2002, 14)
lbr2 = libro("22.11.63", "Stephen King", 1994, 3)
lbr3 = libro("Atti Umani", "Han Kang", 2012, 5)

prestiti = []

prestiti.append(presta_libro(utn1, lbr1, 14))
prestiti.append(presta_libro(utn2, lbr2, 22))
prestiti.append(presta_libro(utn3, lbr3, 3))

for i in prestiti:
    i.dettagli()

print("COPIE DISPONIBILI")
libreria = [lbr1, lbr2, lbr3]

for libro in libreria:
    libro.info()


        
```

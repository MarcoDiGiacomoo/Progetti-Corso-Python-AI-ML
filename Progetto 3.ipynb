```python
#*Parte 1 – Variabili e Tipi di Dati*
#Definisci variabili per rappresentare le seguenti informazioni di un cliente:
#1. nome (stringa)
#2. età (intero)
#3. saldo conto (float)
#4. stato VIP (booleano)
#Esempio: nome = "Mario Rossi", eta = 34, saldo = 2500.75, vip = True
#5. Crea una lista di destinazioni disponibili (almeno 5 città). Definisci un dizionario che associa ogni destinazione a un prezzo medio del viaggio.

nome = "Marco"
eta = 26
saldo_conto = 3896.89
stato_vip = True
destinazioni = ["Roma", "New York", "Lisbona", "Tokyo", "Honk Kong", "Ha Noi", "New Orleans", "Los Angeles", "Pechino"]

prezzi = {destinazioni[0]: 80,
          destinazioni[1]: 550,
          destinazioni[2]: 220,
          destinazioni[3]: 800,
          destinazioni[4]: 700,
          destinazioni[5]: 850,
          destinazioni[6]: 700,
          destinazioni[7]: 650,
          destinazioni[8]: 650}

print(f"Nome: {nome}, Età: {eta}, Saldo: {saldo_conto}, Stato Vip: {stato_vip}, Elenco Destinazioni: {destinazioni}")

for destinazione, prezzo in prezzi.items():
    print(f"Destinazione: {destinazione}, Prezzo: {prezzo}")
```

```python
#*Parte 2 – Programmazione ad Oggetti (OOP)*
#- Crea una classe Cliente con attributi: nome, età, vip.
#- Aggiungi un metodo per stampare le informazioni.
#- Crea una classe Viaggio con attributi: destinazione, prezzo, durata in giorni.
#- Crea una classe Prenotazione che colleghi un cliente a un viaggio. Deve calcolare l’importo finale, con sconto del 10% se il cliente è VIP.
#- Aggiungi un metodo dettagli() che stampa le informazioni complete.

class Cliente:
    def __init__(self, nome, eta, vip):
        self.nome = nome
        self.eta = eta
        self.vip = vip

    def stampa_info(self):
        print(f"Nome: {self.nome}, Eta: {self.eta}, Stato Vip: {self.vip}")

class Viaggio:
    def __init__(self, destinazione, prezzo, durata_giorni):
        self.destinazione = destinazione
        self.prezzo = prezzo
        self.durata_giorni = durata_giorni

class Prenotazione:
    def __init__(self, cliente, viaggio):
        self.cliente = cliente
        self.viaggio = viaggio

    def calcola_prezzo(self):
        if self.cliente.vip != True:
            prezzo = self.viaggio.prezzo
            return prezzo
        else:
            prezzo = self.viaggio.prezzo - (self.viaggio.prezzo * 0.1)
            return prezzo

    def stampa(self):
        print(f"Nome: {self.cliente.nome}, Viaggio: {self.viaggio.destinazione}, Prezzo: {self.calcola_prezzo}")
        
```

```python
#*Parte 3 – NumPy*
#- Genera un array NumPy di 100 prenotazioni simulate, con prezzi casuali fra 200 e 2000 €.
#- Calcola e stampa: prezzo medio, prezzo minimo e massimo, deviazione standard, percentuale di prenotazioni sopra la media.
import numpy as np
import pandas as pd

np.random.seed(41)

prenotazioni = np.random.randint(200, 2000, 100)

media = prenotazioni.mean()
minimo = prenotazioni.min()
massimo = prenotazioni.max()
dev_std = prenotazioni.std()

print(f"Media: {media}")
print(f"Minimo: {minimo}")
print(f"Massimo: {massimo}")
print(f"Deviazione Standard: {dev_std}")

sopra_media = np.sum(prenotazioni > media)
perc = (sopra_media/len(prenotazioni)) * 100

print(f"Percentuale Prenotazioni sopra la media: {perc} %")



#*Parte 4 – Pandas*
#- Crea un DataFrame Pandas con colonne: Cliente, Destinazione, Prezzo, Giorno_Partenza, Durata, Incasso.
#- Calcola con Pandas: incasso totale dell’agenzia, incasso medio per destinazione, top 3 destinazioni più vendute.

df = pd.DataFrame({
    "Cliente": ["Mario Rossi", "Anna Verdi", "Luca Neri", "Giulia Bianchi", "Marco Esposito", "Sara Romano", "Paolo Conti", "Elena Ricci", "Davide Greco", "Chiara Fontana"],
    "Destinazione": ["Parigi", "Roma", "Londra", "Parigi", "Madrid", "Roma", "Parigi", "Berlino", "Londra", "Roma"],
    "Prezzo": [850, 450, 1200, 780, 950, 520, 890, 730, 1150, 480],
    "Giorno_Partenza": ["2026-06-10", "2026-06-15", "2026-07-01", "2026-07-10", "2026-07-20", "2026-08-05", "2026-08-15", "2026-08-22", "2026-09-01", "2026-09-12"],
    "Durata": [5, 3, 7, 4, 6, 3, 5, 4, 8, 3],
    "Incasso": [850, 450, 1200, 780, 950, 520, 890, 730, 1150, 480]
})

incasso = df["Incasso"].sum()
print()
print(f"Incasso Totale: {incasso}")

des_incasso = df.groupby("Destinazione")["Incasso"].mean()
print()
print(f"Incasso per Destinazione: {des_incasso}")

des_piu_vendute = df["Destinazione"].value_counts().head(3)
print()
print(f"Destinazioni più vendute: {des_piu_vendute}")
```

```python
#*Parte 5 – Matplotlib*
#- Crea un grafico a barre che mostri l’incasso per ogni destinazione.

import matplotlib.pyplot as plt

plt.bar(df["Destinazione"], df["Incasso"], color = "lightblue")
plt.xlabel("Destinazioni")
plt.ylabel("Incassi")
plt.show()
```

```python
#- Crea un grafico a linee che mostri l’andamento giornaliero degli incassi.
import matplotlib.dates as mdates

mdates
plt.plot(df["Giorno_Partenza"], df["Incasso"], marker = "o")
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter("%d %m"))
plt.gca().xaxis.set_major_locator(mdates.DayLocator(interval=2))
plt.xlabel("Giorni")
plt.ylabel("incasso")
plt.show()
```

```python
#- Crea un grafico a torta che mostri la percentuale di vendite per ciascuna destinazione.

plt.pie(des_incasso, labels = des_incasso.index, autopct = '%1.1f%%')
plt.show()
```

```python
#*Parte 6 – Analisi Avanzata*
#- Raggruppa i viaggi in categorie: "Europa", "Asia", "America", "Africa". (Puoi usare un dizionario che associa ogni destinazione a una categoria).
#- Calcola con Pandas: incasso totale per categoria, durata media dei viaggi per categoria. Salva il DataFrame aggiornato in un CSV chiamato prenotazioni_analizzate.csv.

df["Categoria"] = ["Europa Occidentale", 
                   "Europa Meridionale", 
                   "Europa Occidentale", 
                   "Europa Occidentale", 
                   "Europa Meridionale", 
                   "Europa Meridionale", 
                   "Europa Occidentale", 
                   "Europa Centrale", 
                   "Europa Occidentale", 
                   "Europa Meridionale"]

#*Parte 7 – Estensioni*
#- Crea una funzione che restituisce i N clienti con più prenotazioni.
#- Realizza un grafico combinato (barre + linea) che mostri: barre = incasso medio per categoria, linea = durata media per categoria.

def clienti_prenotazioni(df, n):
    return df["Cliente"].value_counts().head(n)

clienti_prenotazioni(df, 4)

fig, ax = plt.subplots(1, 2)
ax[0].bar(df["Categoria"], df["Incasso"], color ="lightblue")
ax[0].tick_params(axis = "x", rotation = 45)
durata_media = df.groupby("Categoria")["Durata"].mean()
ax[1].plot(durata_media.index, durata_media.values)
ax[1].tick_params(axis ="x", rotation =45)
plt.show()

```
